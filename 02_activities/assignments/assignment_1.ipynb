### Part 1: Building the base Anagram Checker

Given two valid strings, check to see if they are anagrams of each other. If it is, return `True`, else `False`. For this part, we can assume that uppercase letters are the same as if it was a lowercase character.

Examples of anagrams:
* Silent and Listen
* Night and Thing

Example outputs:
```python
anagram_checker("Silent", "listen") # True
anagram_checker("Silent", "Night") # False
anagram_checker("night", "Thing") # True
```

````python
# For testing purposes, we will write our code in the function

# For better modularity, organization, and functionality, I will divide the program, into two steps, checking for:
# 1. Exact Letter Count - I will check if the two words have the same number of letters.
# 2. Letter Frequency - I will check if the letters in both words appear with the same frequency.


# 1.Exact Letter Count function- I will check if the two words have the same number of letters.
def letter_count(word_a, word_b):
  
  """
  Check if two words have the same number of letters.
  
  Parameters:
  word_a (str): The first word to compare.
  word_b (str): The second word to compare.
  
  Returns:
  bool: True if the words have the same number of letters, False otherwise.
  
  Example:
  >>> letter_count("hello", "world")
  True
  >>> letter_count("python", "java")
  False
  
  """

  return len(word_a) == len(word_b)
````

````python
# 2. Letter Frequency function - I will check if the letters in both words appear with the same frequency.
def letter_frequency(word_a, word_b):
  
  """
  Check if the letters in both words appear with the same frequency.
  
  Parameters:
  word_a (str): The first word to compare.
  word_b (str): The second word to compare.
  
  Returns:
  bool: True if the letters in both words appear with the same frequency, False otherwise.
  
  Example:
  >>> letter_frequency("silent", "listen")
  True
  >>> letter_frequency("python", "java")
  False
  
  """

  from collections import Counter
  
  return Counter(word_a) == Counter(word_b)
````

````python
# Using the two functions above, I will now write the main function to check if two words are anagrams of each other.
def anagram_checker(word_a, word_b):
  
  """
  Check if two words are anagrams of each other.
  
  Parameters:
  word_a (str): The first word to compare.
  word_b (str): The second word to compare.
  
  Returns:
  bool: True if the words are anagrams, False otherwise.
  
  Example:
  >>> anagram_checker("Silent", "Listen")
  True
  >>> anagram_checker("Python", "Java")
  False
  
  """

  # For this part, we can assume that uppercase letters are the same as if it was a lowercase character.
  word_a = word_a.lower()
  word_b = word_b.lower()

  return letter_count(word_a, word_b) and letter_frequency(word_a, word_b)
````

````python
# Run your code to check using the words below:

# Test 1
anagram_checker("Silent", "listen") # True
````

````python
# Test 2
anagram_checker("Silent", "Night") # False
````

````python
# Test 3
anagram_checker("night", "Thing") # True
````

### Part 2: Expanding the functionality of the Anagram Checker

Using your existing and functional anagram checker, let's add a boolean option called `is_case_sensitive`, which will return `True` or `False` based on if the two compared words are anagrams and if we are checking for case sensitivity.

````python
def anagram_checker(word_a, word_b, is_case_sensitive):
  """
  Check if two words are anagrams of each other.
  
  Parameters:
  word_a (str): The first word to compare.
  word_b (str): The second word to compare.
  is_case_sensitive (bool): Whether the comparison should be case sensitive or not.
  
  Returns:
  bool: True if the words are anagrams, False otherwise.
  
  Example:
  >>> anagram_checker("Silent", "Listen", False)
  True
  >>> anagram_checker("Silent", "Listen", True)
  False
  
  """
  # Check if the comparison should be case sensitive or not. If not, convert both words to lowercase.
  if not is_case_sensitive:
    word_a = word_a.lower()
    word_b = word_b.lower()

  return letter_count(word_a, word_b) and letter_frequency(word_a, word_b)
````

````python
# Run your code to check using the words below:

# Test 1
anagram_checker("Silent", "listen", False) # True
````

````python
# Test 2
anagram_checker("Silent", "listen", True) # False
````

````python
# Test 3
anagram_checker("Silent", "Listen", True) # False
````
